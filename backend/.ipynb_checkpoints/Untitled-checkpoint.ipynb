```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn import preprocessing

data= pd.read_csv("health_fitness_dataset.csv")
health=pd.DataFrame(data)
health
```

```python
health.isnull().sum()
```

```python
exercise_mean=health["exercise_minutes"].mean()
steps_mean=health["steps"].mode()[0]
food_calories_mean=health["food_calories"].mode()[0]
sleep_hours_mean=health["sleep_hours"].mode()[0]
weight_mean=health["weight"].mean()
health=health.fillna(value={"exercise_minutes":exercise_mean,"steps":steps_mean,"food_calories":food_calories_mean,"sleep_hours":sleep_hours_mean,"weight":weight_mean})
health.isnull().sum()
```

```python
# Remove spaces

health.columns = health.columns.str.strip()

# Lowercase text

health['Gender'] = health['Gender'].str.lower()
```

```python
#handling neghative valuesss

numeric_cols = health.select_dtypes(include=['int64','float64']).columns

for col in numeric_cols:
    df = health[health[col] >= 0]
```

```python
Q1 = health.quantile(0.25)
Q3 = health.quantile(0.75)
IQR = Q3 - Q1

# Remove outliers

df = df[~((df < (Q1 - 1.5 * IQR)) |
          (df > (Q3 + 1.5 * IQR))).any(axis=1)]
```
